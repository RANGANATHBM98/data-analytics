```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
print(sns.get_dataset_names())
```

```python
df=sns.load_dataset("tips")
```

```python
df.head()
```

```python
df.tail()
```

```python
sns.relplot(x='total_bill', y='tip', data=df)
```

```python
l = ['red', 'black']

sns.relplot(x='total_bill', y='tip', hue='sex', data=df, palette=l)
```

```python
c = ['green', 'yellow']
sns.relplot(x='total_bill', y='tip', hue='smoker', data=df, palette=c)
```

```python
sns.relplot(x='total_bill',y='tip',hue='smoker',data=df,style='time')
```

```python
sns.relplot(x='total_bill',y='tip',data=df,kind='line')
```

```python
np.random.seed(0) #42
df1=pd.DataFrame(dict(time=np.arange(500),value=np.random.randn(500).cumsum()))
df1.head()
```

```python
sns.relplot(x='time',y='value',data=df1,kind='line')
```

```python
sns.relplot(x='time', y='value', kind='line', data=df1)
plt.show()
```

```python
sns.relplot(x='time', y='value', data=df1, kind='line', sort=False)
plt.show()
```
