```python
import sys
cube=[i**3 for i in range(100)]
print("memory allocated for list is",sys.getsizeof(cube)*len(cube))
```

# np.linspace: this funct is similar to arange() function.in this function instead of step size the number of evenly spaced values between interval is spcified 
1)start:start value of sequence                                    
2)stop end val of sequence                                  
3)num:no of evenly spaced sample to be generated default is 50                   
4)retstep=if true returns sample and step between consecutive number

```python
new_g=np.linspace(1,5,retstep=True)
print(new_g)
print(len(new_g))
```

## generate 16 values and by using array and using the methods 2*8 ,8*2 , 4*4 

```python
arr_1d = np.arange(16)

arr_2_8 = arr_1d.reshape(2, 8)

arr_8_2 = arr_1d.reshape(8, 2)

arr_4_4 = arr_1d.reshape(4, 4)

print("2x8 Array:\n", arr_2_8)
print("\n8x2 Array:\n", arr_8_2)
print("\n4x4 Array:\n", arr_4_4)
```
