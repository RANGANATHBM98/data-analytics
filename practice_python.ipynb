## program that prompts the user for two number and display the addition,subtraction,multiplication and division between them

```python
# prompt the user for two number
num1=float(input("enter the first number"))
num2=float(input("enter the first number"))

#perform arithmetic operation
add=num1+num2
sub=num1-num2
div=num1/num2
mul=num1*num2
#display operation
print("addition is:",add)
print("substraction is:",sub)
print("division is:",div)
print("multiplication is:",mul)

#all
print("Addition:", add, "Subtraction:", sub, "Division:", div, "Multiplication:", mul)
```

## find the geometrical mean of three numbers entered by the user

```python
import math
```

```python
num1=float(input("enter the first number"))
num2=float(input("enter the second number"))
num3=float(input("enter the third number"))
product=num1*num2*num3
a_mean=math.pow(product,1/3)
print("the value of geometric mean is:",a_mean)
```

## wap that display even numbers 1 to 50 and odd number 51 to 100 using repeating loop

```python
num = 1

while num <= 100:
    if num <= 50:
        if num % 2 == 0:
            print(num, "even no")
        else:
            print(num, "odd no")
    else:
        if num % 2 != 0:
            print(num, "odd no")

    num += 1
```
